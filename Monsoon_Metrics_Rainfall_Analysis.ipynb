```python
def classify_rainfall(deviation):
    if pd.isna(deviation):
        return "Unknown"
    elif deviation < -19:
        return "Below Normal"
    elif deviation <= 19:
        return "Normal"
    else:
        return "Above Normal"

df["rainfall_status"] = df["deviation"].apply(classify_rainfall)
```

```python
df["rainfall_status"].value_counts()
```

```python
df["rainfall_status"].value_counts(normalize=True).mul(100).round(2)
```

```python
yearly_status = (
    df.groupby(["year", "rainfall_status"])
      .size()
      .reset_index(name="days")
)

yearly_status
```

```python
yearly_status_pivot = yearly_status.pivot(
    index="year",
    columns="rainfall_status",
    values="days"
).fillna(0)

yearly_status_pivot
```

```python
yearly_status_pivot.plot(
    kind="bar",
    stacked=True,
    figsize=(14, 6)
)

plt.title("Rainfall Status by Year")
plt.xlabel("Year")
plt.ylabel("Number of Observations")
plt.legend(title="Rainfall Status")
plt.tight_layout()
plt.show()
```
